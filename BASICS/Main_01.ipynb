```python
# Reverse number.

a = int(input("Enter a number : "))
b = a
rev = 0

while b>0:
    rev =rev*10 + b%10
    b = b//10

if rev == a:
    print("Number is palindrome")
else:
    print("Number is not palindrome")
```

```python
# Palindrome
```

```python
def palindrome(a):
    b = a
    rev = 0

    while b>0:
        rev =rev*10 + b%10
        b = b//10

    if rev == a:
        print("Number is palindrome")
    else:
        print("Number is not palindrome")

# palindrome(int(input("Enter number : "))) # They will aslo work.
a = int(input("Enter number : "))
palindrome(a)
```

```python
# Positional arguments

def multiply(a,b,c,d):
    print(f"Result : {a*b*c*d}")

multiply(1,2,3,4)
```

```python
# Default argument.

def add(a,b,c=12):
    print(f"Result : {a+b+c}")

add(1,2)  # In this situtiation they will choose default value.
add(1,2,3) # In this case they can choose own value
```

```python
# Keyword argument

def sub(a,b):
    print(f"Result : {a-b}")

sub(b=10, a = 33)
```

```python
# Tuple -> ()
# Ordered, Unmutable, Indexed Access, Allow Duplication.
# a[1] = 2 tries to modify the second element of the tuple, but tuples are read-only.
```

```python
a = (1,2,3,44,44)
print(a)
print(type(a))
```

```python
# Set -> {}
# UnOrdered, mutable, Not Indexed Access, Not Allow Duplication.
```

```python
a = {1,2,3.4,5}
print(a)
print(type(a))
```

```python
# | and {**dict1, **dict2} create a new merged dictionary.

dict1  = {"Gaurav":"Singh"}
dict2  = {"Maths":"Mathematics"}

dict = dict1 | dict2
dict3 = {**dict1 , **dict2}
print(dict)
print(dict3)
```

```python
# update() modifies the original dictionary.

dict1  = {"Gaurav":"Singh"}
dict2  = {"Maths":"Mathematics"}

dict1.update(dict2)
print(dict1)
```
