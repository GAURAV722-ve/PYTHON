```python
# Comparison Operators in Python
# == -> equal to
# != -> not equal to
# > -> greater than
# < -> less than
# >= -> greater than or equal to
# <= -> less than or equal to
```

```python
a = 10
b = 20

print(a == b)  # Output: False
print(a != b)  # Output: True
print(a > b)   # Output: False
print(a < b)   # Output: True
print(a >= b)  # Output: False
print(a <= b)  # Output: True
```

```python
# Logical Operators in Python
# and -> returns True if both statements are true
# or -> returns True if one of the statements is true
# not -> returns True if the statement is false
```

```python
a=10
b=20
print(a == a and b > a) # Output: True
print(a==b or b > a) # Output: True
print(not(a==b)) # Output: True
```

```python
# Print even and odd number in range seprately.

n = int(input("Enter number : "))
sum_odd=0
sum_even =0
for i in range(1,n+1):
    if(i%2==0):
        sum_even +=i
    else:
        sum_odd +=i
print(f"sum of even number is : {sum_even}")
print(f"sum of Odd number is : {sum_odd}")
```

```python
n = int(input("Enter number: "))
sum = 0

for i in range(1, n // 2 + 1):
    if n % i == 0:
        sum += i

if sum == n:
    print("Number is perfect.")
else:
    print("Number is not perfect.")
```

```python
import random

n = random.randint(1, 100)
while True:
    x = int(input("Guess Number: "))

    if x == n:
        print("Congratulations! You have won!")
        break
    elif x > n:
        print("Wrong guess, go lower.")
    else:
        print("Wrong guess, go higher.")
```
